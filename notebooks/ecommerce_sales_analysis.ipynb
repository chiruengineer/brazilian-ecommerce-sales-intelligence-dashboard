```python
date_columns = [
    "shipping_limit_date",
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]
```

```python
for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")
```

```python
df[date_columns].dtypes
```

```python
numeric_columns = [
    "price",
    "freight_value",
    "payment_value",
    "payment_installments",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

df[numeric_columns].describe().T
```

```python
df["revenue"] = df["payment_value"]
```

```python
df["order_year"] = df["order_purchase_timestamp"].dt.year
```

```python
df["order_month"] = df["order_purchase_timestamp"].dt.month
```

```python
df["order_month_name"] = df["order_purchase_timestamp"].dt.month_name()
```

```python
df["order_quarter"] = (
    "Q" + df["order_purchase_timestamp"].dt.quarter.astype(str)
)
```

```python
df["year_month"] = (
    df["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
df["delivery_days"] = (
    df["order_delivered_customer_date"]
    - df["order_purchase_timestamp"]
).dt.days
```

```python
df["delivery_delay_days"] = (
    df["order_delivered_customer_date"]
    - df["order_estimated_delivery_date"]
).dt.days
```

```python
total_revenue = df["revenue"].sum()

print(f"Total Revenue: {total_revenue:,.2f}")
```

```python
total_orders = df["order_id"].nunique()

print(f"Total Orders: {total_orders:,}")
```

```python
total_customers = df["customer_unique_id"].nunique()

print(f"Total Customers: {total_customers:,}")
```

```python
total_products = df["product_id"].nunique()

print(f"Total Products: {total_products:,}")
```

```python
average_order_value = total_revenue / total_orders

print(f"Average Order Value: {average_order_value:,.2f}")
```

```python
monthly_revenue = (
    df.groupby("year_month")["revenue"]
    .sum()
    .reset_index()
)

monthly_revenue.head()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_revenue["year_month"],
    monthly_revenue["revenue"]
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
category_revenue = (
    df.groupby("product_category_name")["revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(15)
)

category_revenue
```

```python
plt.figure(figsize=(10, 7))

category_revenue.sort_values().plot(kind="barh")

plt.title("Top 15 Product Categories by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Product Category")

plt.tight_layout()
plt.show()
```

```python
state_revenue = (
    df.groupby("customer_state")["revenue"]
    .sum()
    .sort_values(ascending=False)
)

state_revenue
```

```python
plt.figure(figsize=(12, 6))

state_revenue.plot(kind="bar")

plt.title("Revenue by Customer State")
plt.xlabel("State")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
payment_analysis = (
    df.groupby("payment_type")
    .agg(
        Revenue=("revenue", "sum"),
        Orders=("order_id", "nunique")
    )
    .sort_values("Revenue", ascending=False)
)

payment_analysis
```

```python
order_status = (
    df.groupby("order_status")
    .agg(
        Orders=("order_id", "nunique"),
        Revenue=("revenue", "sum")
    )
    .sort_values("Orders", ascending=False)
)

order_status
```

```python
plt.figure(figsize=(9, 5))

order_status["Orders"].plot(kind="bar")

plt.title("Orders by Status")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
top_products = (
    df.groupby("product_id")
    .agg(
        Revenue=("revenue", "sum"),
        Orders=("order_id", "nunique")
    )
    .sort_values("Revenue", ascending=False)
    .head(10)
)

top_products
```

```python
cleaned_df = df.copy()
```

```python
columns_to_remove = [
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

cleaned_df = cleaned_df.drop(
    columns=columns_to_remove,
    errors="ignore"
)
```

```python
from pathlib import Path

output_path = Path("../data/cleaned")

output_path.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
cleaned_df.to_csv(
    output_path / "ecommerce_sales_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
